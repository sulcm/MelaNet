```python
cached_features_base = CacheManager[FeatureExtractorCache].load(cached_features_base_path)
```

```python
base_metadata = cached_features_base.metadata
index_to_merge = [cached_features_base.cache["index"],]
queries_to_merge = [cached_features_base.cache["eval_dataset"],]

for add_cache_p in add_cached_features_paths:
    _to_add_cache = CacheManager[FeatureExtractorCache].load(add_cache_p)
    index_to_merge.append(_to_add_cache.cache["index"])
    queries_to_merge.append(_to_add_cache.cache["eval_dataset"])
```

```python
merged_index_featrues = pd.concat(index_to_merge, ignore_index=True)
merged_queries_featrues = pd.concat(queries_to_merge, ignore_index=True)
```
