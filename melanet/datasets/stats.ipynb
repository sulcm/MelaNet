```python
import matplotlib.pyplot as plt

from collections import Counter

from datasets import load_from_disk

from display_utils import draw_pie_chart_with_leader_lines, sample_dataset_and_display
```

```python
dset = load_from_disk("/home/sulcm/datasets/MelanoMix")

CLASS_LABEL_COLUMN = "label"
FEATURE_COLUMN = "image"
splits = list(dset.keys())

dset
```

```python
split_counts = {s: Counter(dset[s][CLASS_LABEL_COLUMN]) for s in splits}
overall_counts = sum([c for c in split_counts.values()], start=Counter())
```

```python
named_split_counts = {
    s: {
        dset[s].features[CLASS_LABEL_COLUMN].int2str(k): v
        for k, v in
        split_counts[s].items()
    }
    for s in splits
}
named_overall_counts = {dset[splits[0]].features[CLASS_LABEL_COLUMN].int2str(k): v for k, v in overall_counts.items()}
```

```python
fig, ax = plt.subplots(figsize=(11, 7))
draw_pie_chart_with_leader_lines(
    ax,
    list(named_overall_counts.values()),
    list(named_overall_counts.keys()),
    popout_percentage=True,
    add_legend=True
)
plt.title("Dataset class distribution")
plt.show()
```
