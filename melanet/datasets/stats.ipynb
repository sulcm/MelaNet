```python
import matplotlib.pyplot as plt
from matplotlib.axes import Axes

from collections import Counter

from datasets import load_from_disk

from display_utils import draw_pie_chart_with_leader_lines, sample_dataset_and_display
```

```python
# dset = load_from_disk("/home/sulcm/datasets/isic2019/isic2019")
dset = load_from_disk("/home/sulcm/datasets/ham10000/HAM10000")
# dset = load_from_disk("/home/sulcm/datasets/milk10k/milk10k")

CLASS_LABEL_COLUMN = "label"
FEATURE_COLUMN = "image"
splits = list(dset.keys())

dset
```

```python
def get_split_count(dset, splits: list[str], named: bool = False):
    _split_counts = {s: Counter(dset[s][CLASS_LABEL_COLUMN]) for s in splits}

    if named:
        split_counts = {
            s: {
                dset[s].features[CLASS_LABEL_COLUMN].int2str(k): v
                for k, v in
                _split_counts[s].items()
            }
            for s in splits
        }
    else:
        split_counts = _split_counts

    return split_counts
```

```python
split_counts = get_split_count(dset, splits)
named_split_counts = get_split_count(dset, splits,named=True)
```

```python
overall_counts = sum([c for c in split_counts.values()], start=Counter())
named_overall_counts = {dset[splits[0]].features[CLASS_LABEL_COLUMN].int2str(k): v for k, v in overall_counts.items()}
```

```python
fig, ax = plt.subplots(1, len(splits), figsize=(len(splits)*5, 3), sharey=True)
sorted_labels = [dset["train"].features[CLASS_LABEL_COLUMN].int2str(l) for l, c in split_counts["train"].most_common()]
global_max_count = 0
for i, s in enumerate(splits):
    s_counts = named_split_counts[s]
    _ax = ax if isinstance(ax, Axes) else ax[i]
    rects = _ax.bar(
        list(range(len(sorted_labels))),
        [s_counts.get(l, 0.0) for l in sorted_labels],
        label=sorted_labels
    )
    _ax.bar_label(rects, padding=1)
    _ax.set_title(f"{s.capitalize()} Split")
    # if s == "train":
    #     _ax.set_title(f"{s.capitalize()} Split")
    # else:
    #     _ax.set_title(f"{s.capitalize()} + Test Split")
    _ax.set_xticks(list(range(len(sorted_labels))), sorted_labels, rotation=45)
    _curr_max_count = max(s_counts.values())
    if global_max_count < _curr_max_count:
        global_max_count = _curr_max_count
    # _ax.set_ylim(0, max(s_counts.values()) * 1.1)
    if i == 0:
        _ax.set_ylabel("Number of samples")

for i in range(len(splits)):
    _ax = ax if isinstance(ax, Axes) else ax[i]
    _ax.set_ylim(0, global_max_count * 1.1)

plt.tight_layout()
# plt.savefig("/home/sulcm/school/ST/DP/images/milk10k_histogram.pdf", bbox_inches="tight")
# plt.savefig("/home/sulcm/school/ST/DP/images/ham10000_histogram.pdf", bbox_inches="tight")
# plt.savefig("/home/sulcm/school/ST/DP/images/isic2019_histogram.pdf", bbox_inches="tight")
plt.show()
```

```python
isic2019_dset = load_from_disk("/home/sulcm/datasets/isic2019/isic2019")
ham_dset = load_from_disk("/home/sulcm/datasets/ham10000/HAM10000")
milk_dset = load_from_disk("/home/sulcm/datasets/milk10k/milk10k")

CLASS_LABEL_COLUMN = "label"
FEATURE_COLUMN = "image"
COMP_SPLIT = "train"
splits = [COMP_SPLIT,]
```

```python
comp_datasets = [(milk_dset, "MILK10k"), (isic2019_dset, "ISIC 2019"), (ham_dset, "ISIC 2018 (HAM10000)")]
# fig, ax = plt.subplots(1, len(comp_datasets), figsize=(len(comp_datasets)*4.5, 3), sharey=True)
# Create the main figure window
fig = plt.figure(figsize=(10, 7))

# Define a 2x2 grid layout
gs = fig.add_gridspec(2, 2)

# First Row: Two separate plots side-by-side
ax1 = fig.add_subplot(gs[0, 0])  # Row 0, Column 0
ax2 = fig.add_subplot(gs[0, 1])  # Row 0, Column 1

# Second Row: One plot spanning both columns [row, col_start:col_end]
ax3 = fig.add_subplot(gs[1, :])  # Row 1, spans Column 0 and 1

ax = [ax3, ax1, ax2]

global_max_count = 0
for i, dset_tuple in enumerate(comp_datasets):
    dset, dset_name = dset_tuple
    split_counts = get_split_count(dset, splits)
    named_split_counts = get_split_count(dset, splits, named=True)
    sorted_labels = [dset[COMP_SPLIT].features[CLASS_LABEL_COLUMN].int2str(l) for l, c in split_counts[COMP_SPLIT].most_common()]
    s_counts = named_split_counts[COMP_SPLIT]

    _ax = ax if isinstance(ax, Axes) else ax[i]
    rects = _ax.bar(
        list(range(len(sorted_labels))),
        [s_counts.get(l, 0.0) for l in sorted_labels],
        label=sorted_labels
    )
    _ax.bar_label(rects, padding=1)
    _ax.set_title(dset_name)
    _ax.set_xticks(list(range(len(sorted_labels))), sorted_labels, rotation=45)
    _curr_max_count = max(s_counts.values())
    if global_max_count < _curr_max_count:
        global_max_count = _curr_max_count
    if i == 0 or i == 1:
        _ax.set_ylabel("Number of samples")

for i in range(len(comp_datasets)):
    _ax = ax if isinstance(ax, Axes) else ax[i]
    _ax.set_ylim(0, global_max_count * 1.1)

plt.tight_layout()
# plt.savefig("/home/sulcm/school/ST/DP/images/milk10k_dataset_hist_comp.pdf", bbox_inches="tight")
plt.show()
```
